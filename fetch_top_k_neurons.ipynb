```python
import json 

data = json.load(open("data/sample.json"))

new_data = {}
for k, v in data.items():
    new_data[k] = {}
    for k2, v2 in v.items():
        new_data[k][k2] = []
        for i in range(2):
            new_data[k][k2].append(v2[i])

print(len(new_data))
json.dump(new_data, open("data/sample_2.json", "w"), ensure_ascii=False, indent=4)
```
